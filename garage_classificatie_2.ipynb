```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.metrics import accuracy_score, precision_score, recall_score
```

# Data laden
df = pd.read_excel('AmesHousing.xlsx', sheet_name='AmesHousing')
df['Garage'] = df['Garage Type'].apply(lambda x: 'no' if pd.isna(x) else 'yes')

```python
df1 = pd.get_dummies(df, columns=['Neighborhood'])

neighborhood_cols = [c for c in df1.columns if 'Neighborhood' in c]
features1 = ['Gr Liv Area', 'Year Built'] + neighborhood_cols

X1 = df1[features1]
y1 = df1['Garage']

X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size=0.15, random_state=42)

model1 = DecisionTreeClassifier(max_depth=4, min_samples_split=20, random_state=42)
model1.fit(X1_train, y1_train)

y_pred1 = model1.predict(X1_test)

acc1   = accuracy_score(y1_test, y_pred1)
prec1  = precision_score(y1_test, y_pred1, pos_label='yes')
rec1   = recall_score(y1_test, y_pred1, pos_label='yes')

print('accuracy: ', acc1)
print('precision:', prec1)
print('recall:   ', rec1)
```

```python
df3 = pd.get_dummies(df, columns=['Neighborhood', 'House Style'])
neighborhood_cols = [c for c in df3.columns if 'Neighborhood' in c]
housestyle_cols = [c for c in df3.columns if 'House Style' in c]
features3 = ['Gr Liv Area', 'Year Built', 'Lot Area', 'Full Bath', 'Overall Qual', 'Total Bsmt SF'] + neighborhood_cols + housestyle_cols

X3 = df3[features3].fillna(0)
y3 = df3['Garage']
X3_train, X3_test, y3_train, y3_test = train_test_split(X3, y3, test_size=0.15, random_state=42)

model3 = DecisionTreeClassifier(max_depth=5, min_samples_split=15, min_samples_leaf=5, random_state=42)
model3.fit(X3_train, y3_train)

y_pred3 = model3.predict(X3_test)
acc3 = accuracy_score(y3_test, y_pred3)
prec3 = precision_score(y3_test, y_pred3, pos_label='yes')
rec3 = recall_score(y3_test, y_pred3, pos_label='yes')

tree_rules3 = export_text(model3, feature_names=features3, class_names=['no', 'yes'])
print(tree_rules3)
print(f"Acc: {acc3:.1%} | Prec: {prec3:.1%} | Rec: {rec3:.1%}")
```
