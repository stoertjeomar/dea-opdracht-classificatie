```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score, precision_score, recall_score
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('AmesHousing.xlsx', sheet_name='AmesHousing')

# maak een kolom Garage: yes als er een garage type is, anders no
df['Garage'] = df['Garage Type'].apply(lambda x: 'no' if pd.isna(x) else 'yes')

df.head()
```

## Experiment 1
Simpele features, kleine boom

```python
df1 = pd.get_dummies(df, columns=['Neighborhood'])

neighborhood_cols = [c for c in df1.columns if 'Neighborhood' in c]
features1 = ['Gr Liv Area', 'Year Built'] + neighborhood_cols

X1 = df1[features1]
y1 = df1['Garage']

X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size=0.15, random_state=42)

model1 = DecisionTreeClassifier(max_depth=4, min_samples_split=20, random_state=42)
model1.fit(X1_train, y1_train)

y_pred1 = model1.predict(X1_test)

acc1   = accuracy_score(y1_test, y_pred1)
prec1  = precision_score(y1_test, y_pred1, pos_label='yes')
rec1   = recall_score(y1_test, y_pred1, pos_label='yes')

print('accuracy: ', acc1)
print('precision:', prec1)
print('recall:   ', rec1)
```

```python
plt.figure(figsize=(20, 8))
plot_tree(model1, feature_names=features1, class_names=['no', 'yes'], filled=True)
plt.title('Experiment 1')
plt.show()
```

```python
cm1 = confusion_matrix(y1_test, y_pred1, labels=['no', 'yes'])
ConfusionMatrixDisplay(cm1, display_labels=['no', 'yes']).plot()
plt.title('Experiment 1')
plt.show()
```

## Experiment 2
Meer features toegevoegd, diepere boom

```python
df2 = pd.get_dummies(df, columns=['Neighborhood', 'House Style'])

neighborhood_cols = [c for c in df2.columns if 'Neighborhood' in c]
housestyle_cols   = [c for c in df2.columns if 'House Style' in c]
features2 = ['Gr Liv Area', 'Year Built', 'Lot Area', 'Full Bath'] + neighborhood_cols + housestyle_cols

X2 = df2[features2]
y2 = df2['Garage']

X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.15, random_state=42)

model2 = DecisionTreeClassifier(max_depth=6, min_samples_leaf=10, random_state=42)
model2.fit(X2_train, y2_train)

y_pred2 = model2.predict(X2_test)

acc2   = accuracy_score(y2_test, y_pred2)
prec2  = precision_score(y2_test, y_pred2, pos_label='yes')
rec2   = recall_score(y2_test, y_pred2, pos_label='yes')

print('accuracy: ', acc2)
print('precision:', prec2)
print('recall:   ', rec2)
```

```python
cm2 = confusion_matrix(y2_test, y_pred2, labels=['no', 'yes'])
ConfusionMatrixDisplay(cm2, display_labels=['no', 'yes']).plot()
plt.title('Experiment 2')
plt.show()
```

## Experiment 3
Nog meer features, extra hyperparameters

```python
df3 = pd.get_dummies(df, columns=['Neighborhood', 'House Style'])

neighborhood_cols = [c for c in df3.columns if 'Neighborhood' in c]
housestyle_cols   = [c for c in df3.columns if 'House Style' in c]
features3 = ['Gr Liv Area', 'Year Built', 'Lot Area', 'Full Bath', 'Overall Qual', 'Total Bsmt SF'] + neighborhood_cols + housestyle_cols

X3 = df3[features3].fillna(0)
y3 = df3['Garage']

X3_train, X3_test, y3_train, y3_test = train_test_split(X3, y3, test_size=0.15, random_state=42)

model3 = DecisionTreeClassifier(max_depth=5, min_samples_split=15, min_samples_leaf=5, random_state=42)
model3.fit(X3_train, y3_train)

y_pred3 = model3.predict(X3_test)

acc3   = accuracy_score(y3_test, y_pred3)
prec3  = precision_score(y3_test, y_pred3, pos_label='yes')
rec3   = recall_score(y3_test, y_pred3, pos_label='yes')

print('accuracy: ', acc3)
print('precision:', prec3)
print('recall:   ', rec3)
```

```python
cm3 = confusion_matrix(y3_test, y_pred3, labels=['no', 'yes'])
ConfusionMatrixDisplay(cm3, display_labels=['no', 'yes']).plot()
plt.title('Experiment 3')
plt.show()
```

## Experiment 4
Alle features, ondiepe boom om overfitten te voorkomen

```python
df4 = pd.get_dummies(df, columns=['Neighborhood', 'House Style'])

neighborhood_cols = [c for c in df4.columns if 'Neighborhood' in c]
housestyle_cols   = [c for c in df4.columns if 'House Style' in c]
features4 = ['Gr Liv Area', 'Year Built', 'Lot Area', 'Full Bath', 'Overall Qual', 'Total Bsmt SF', 'Bedroom AbvGr', 'Garage Cars'] + neighborhood_cols + housestyle_cols

X4 = df4[features4].fillna(0)
y4 = df4['Garage']

X4_train, X4_test, y4_train, y4_test = train_test_split(X4, y4, test_size=0.15, random_state=42)

model4 = DecisionTreeClassifier(max_depth=3, min_samples_split=10, random_state=42)
model4.fit(X4_train, y4_train)

y_pred4 = model4.predict(X4_test)

acc4   = accuracy_score(y4_test, y_pred4)
prec4  = precision_score(y4_test, y_pred4, pos_label='yes')
rec4   = recall_score(y4_test, y_pred4, pos_label='yes')

print('accuracy: ', acc4)
print('precision:', prec4)
print('recall:   ', rec4)
```

```python
cm4 = confusion_matrix(y4_test, y_pred4, labels=['no', 'yes'])
ConfusionMatrixDisplay(cm4, display_labels=['no', 'yes']).plot()
plt.title('Experiment 4')
plt.show()
```

## Overzicht alle experimenten

```python
overzicht = pd.DataFrame({
    'Metric':         ['accuracy', 'precision', 'recall'],
    'Experiment 1':   [round(acc1, 4), round(prec1, 4), round(rec1, 4)],
    'Experiment 2':   [round(acc2, 4), round(prec2, 4), round(rec2, 4)],
    'Experiment 3':   [round(acc3, 4), round(prec3, 4), round(rec3, 4)],
    'Experiment 4':   [round(acc4, 4), round(prec4, 4), round(rec4, 4)],
})

print(overzicht.to_string(index=False))
overzicht
```
