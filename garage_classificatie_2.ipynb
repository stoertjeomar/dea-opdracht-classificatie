```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score, precision_score, recall_score
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('AmesHousing.xlsx', sheet_name='AmesHousing')
df.head()
```

## experiment 1

```python
df1 = pd.get_dummies(df, columns=['Neighborhood'])

neighborhood_cols = [c for c in df1.columns if 'Neighborhood' in c]
features = ['Gr Liv Area', 'Year Built'] + neighborhood_cols

X = df1[features]
y = df1['Garage']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=42)
```

```python
model = DecisionTreeClassifier(max_depth=4, min_samples_split=20, random_state=42)
model.fit(X_train, y_train)
```

```python
plt.figure(figsize=(20, 8))
plot_tree(model, feature_names=features, class_names=['no', 'yes'], filled=True)
plt.show()
```

```python
y_pred = model.predict(X_test)

print('accuracy:', accuracy_score(y_test, y_pred))
print('precision:', precision_score(y_test, y_pred, pos_label='yes'))
print('recall:', recall_score(y_test, y_pred, pos_label='yes'))
```

```python
cm = confusion_matrix(y_test, y_pred, labels=['no', 'yes'])
ConfusionMatrixDisplay(cm, display_labels=['no', 'yes']).plot()
plt.show()
```

## experiment 2

```python
df2 = pd.get_dummies(df, columns=['Neighborhood', 'House Style'])

neighborhood_cols = [c for c in df2.columns if 'Neighborhood' in c]
housestyle_cols = [c for c in df2.columns if 'House Style' in c]
features2 = ['Gr Liv Area', 'Year Built', 'Lot Area', 'Full Bath'] + neighborhood_cols + housestyle_cols

X2 = df2[features2]
y2 = df2['Garage']
```
